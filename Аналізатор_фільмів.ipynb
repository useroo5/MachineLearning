```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import ast
```

```python
url = "movies_metadata.csv"

df.head()
```

```python
def extract_genres(genres_str):
  try:
    genres = ast.literal_eval(genres_str)
    return [genre['name'] for genre in genres]
  except ValueError:
    return[]
```

```python
try:
  df['genres'] = df['genres'].apply(extract_genres)
except:
  df['genres'] = []
```

```python
genre_counts = df['genres'].value_counts()
print(genre_counts)
```

```python
genres_exploded = df[['release_year','title', 'release_date', 'budget', 'revenue', 'genres']].explode('genres')
genres_exploded.head(10)
```

```python
genre_count = genres_exploded['genres'].value_counts()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))

sns.barplot(x=genre_count.index, y=genre_count.values)

plt.title("Кол-во фильмов по жанрам")
plt.xlabel("Жанр")
plt.ylabel("Кол-во фильмов")

plt.xticks(rotation=90)

plt.show()
```

```python
df['popularity'] = pd.to_numeric(df['popularity'], errors='coerce')
popularity_trend = df.groupby('release_year')['popularity'].mean()

plt.figure(figsize=(12,6))
popularity_trend.plot()
plt.title("Середня популярність фільмів за роками")
plt.xlabel("Рік")
plt.ylabel("Популярність")

plt.tight_layout()
plt.show()

```
