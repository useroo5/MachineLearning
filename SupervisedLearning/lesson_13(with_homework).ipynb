```python
import numpy
import pandas as pd

p=True


while p:

  coins_collected = list(map(int, input("How many coins have you collected each day?(Enter using space): ").split()))
  print(coins_collected)

  days = list(map(int, input("What was the time period in days?(Enter using space): ").split()))
  print(days)

  if len(days) != len(coins_collected):
    print("please type an equal amount of days AND coins")
    days.clear
    coins_collected.clear

  else:
    try:
      prediction = [[int(input("Enter a day to predict the amount of coins: "))]]
      p=False

    except:
      print("please enter a number next time")

data = {
    "Days": days,
    "Coins": coins_collected,
}
print(data)
df = pd.DataFrame(data)

df.head()
df.info()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df['Days'], df['Coins'], color = "blue")

plt.title("График зависимости")

plt.xlabel("Days")
plt.ylabel("Coins collected")

plt.show()
```

```python
## scikit-learn
from sklearn.linear_model import LinearRegression

X = df[["Days"]]
Y = df["Coins"]

model = LinearRegression()
model.fit(X, Y)

print("Коэффициент (W): ", model.coef_)
print("Смещение (b): ", model.intercept_)
```

```python
Y_pred = model.predict(X)

plt.scatter(X, Y, color = "blue")
plt.plot(X, Y_pred, color = "red")
plt.title("Линейная регрессия")
plt.xlabel("Days")
plt.ylabel("Coins")

plt.show()
```
